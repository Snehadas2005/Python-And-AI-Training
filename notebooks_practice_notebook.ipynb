# Data Analysis Practice Notebook
This notebook demonstrates how to load a CSV file, analyze data using Pandas, and visualize trends using Matplotlib.

```python
import pandas as pd
import matplotlib.pyplot as plt
from pyodide.http import open_url

# Use pyodide's open_url to fetch data inside the browser
url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
df = pd.read_csv(open_url(url))

# Display the first 5 rows
df.head()
```

```python
# Plot a bar chart showing passenger count by Class
df['Pclass'].value_counts().plot(kind='bar', color='skyblue')
plt.title('Passenger Count by Class')
plt.xlabel('Ticket Class (1 = 1st, 2 = 2nd, 3 = 3rd)')
plt.ylabel('Number of Passengers')
plt.show()
```

```python
# Basic info and summary statistics
df.info()
df.describe()
```

```python
# Check for missing values in each column
df.isnull().sum()
```

```python
# Survival rate by Gender
df.groupby('Sex')['Survived'].mean().plot(kind='bar', color=['lightpink', 'skyblue'])
plt.title('Survival Rate by Gender')
plt.ylabel('Survival Rate')
plt.show()

# Age distribution of passengers
df['Age'].plot(kind='hist', bins=20, edgecolor='black')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.show()
```

## Key Insights & Conclusion from Titanic EDA

### Summary of Findings:
* **Gender and Survival:** Female passengers had a significantly higher survival rate compared to male passengers.
* **Passenger Class:** Passengers traveling in 1st Class had the highest survival rate, whereas 3rd Class passengers accounted for the largest portion of the total passenger count but had lower survival outcomes.
* **Age Distribution:** Most passengers were young adults between the ages of 20 and 40.
* **Missing Data:** 
  * The `Age` column has 177 missing values (which could be imputed using the median age).
  * The `Cabin` column has 687 missing values (most data is missing, so dropping or re-categorizing is recommended).
  * The `Embarked` column has 2 missing values.
